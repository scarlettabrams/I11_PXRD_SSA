## Imports 

```python
# imports & helper functions
import glob
import os

import h5py
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Try importing scipy.signal utilities; fall back to numpy versions if not installed
try:
    from scipy.signal import correlate, find_peaks
except Exception:
    correlate = None
    find_peaks = None
    print(
        "Note: scipy.signal not available - using numpy fallback for cross-correlation. "
        "If you want better peak finding/xcorr, install scipy with `conda install -c conda-forge scipy`."
    )


# small helper: robust text file loader that skips headers
def robust_loadtxt(path, expected_cols=2, max_header_lines=50):
    """
    Load numeric text data while tolerating short text headers.
    Returns an (N, M) numpy array.
    """
    try:
        return np.loadtxt(path)
    except Exception:
        pass

    with open(path, "r", errors="ignore") as f:
        probe_lines = []
        for _ in range(max_header_lines):
            line = f.readline()
            if not line:
                break
            probe_lines.append(line)

    numeric_start = None
    for i, ln in enumerate(probe_lines):
        parts = ln.strip().split()
        if len(parts) < expected_cols:
            continue
        try:
            [float(x) for x in parts[:expected_cols]]
            numeric_start = i
            break
        except Exception:
            continue

    start_try = numeric_start if numeric_start is not None else 0
    for skiprows in range(start_try, max_header_lines + 1):
        try:
            data = np.loadtxt(path, skiprows=skiprows)
            if data.ndim == 1:
                data = data.reshape(-1, len(data))
            return data
        except Exception:
            continue

    raise ValueError(
        f"Could not parse numeric data from {path} with up to {max_header_lines} header lines."
    )


def resolve_fep_input_path(fep_input, collection_no=None):
    """
    Resolve FEP input path.
    - If fep_input is a file path, return it directly.
    - If fep_input is a folder, pick file by collection_no if provided, else first matching I11 prefix file.
    """
    if os.path.isfile(fep_input):
        return fep_input

    if not os.path.isdir(fep_input):
        raise FileNotFoundError(
            f"FEP input not found: {fep_input}. Provide a file path or a valid directory."
        )

    candidates = []
    if collection_no is not None and str(collection_no).strip() != "":
        c = str(collection_no).strip()
        candidates.extend(
            [
                os.path.join(fep_input, f"i11-1-{c}.nxs"),
                os.path.join(fep_input, f"pixium_{c}.hdf"),
                os.path.join(fep_input, f"pixium_{c}.h5"),
            ]
        )

    for c in candidates:
        if os.path.exists(c):
            return c

    # fallback: first matching file by your naming conventions
    patterns = [
        os.path.join(fep_input, "i11-1-*.nxs"),
        os.path.join(fep_input, "pixium_*.hdf"),
        os.path.join(fep_input, "pixium_*.h5"),
    ]
    for pat in patterns:
        found = sorted(glob.glob(pat))
        if found:
            return found[0]

    raise FileNotFoundError(
        "No matching FEP file found in folder. Expected i11-1-*.nxs or pixium_*.hdf/.h5."
    )


def resolve_poni_path(poni_path):
    """
    Accept either:
    - direct .poni file path
    - directory that contains one .poni file
    Also cleans accidental trailing slash after a .poni filename.
    """
    if poni_path is None or str(poni_path).strip() == "":
        return None

    p = os.path.normpath(str(poni_path).strip())

    # If user passed '<file>.poni/' by accident, strip separators and retry as file path
    p_stripped = str(poni_path).strip().rstrip("/\\")
    if p.lower().endswith(".poni") and os.path.isfile(p):
        return p
    if p_stripped.lower().endswith(".poni") and os.path.isfile(p_stripped):
        return p_stripped

    if os.path.isdir(p):
        found = sorted(glob.glob(os.path.join(p, "*.poni")))
        if len(found) == 1:
            return found[0]
        if len(found) > 1:
            raise ValueError(
                f"Multiple .poni files found in {p}. Please provide the exact .poni file path."
            )
        raise FileNotFoundError(f"No .poni file found in directory: {p}")

    raise FileNotFoundError(f".poni path not found: {poni_path}")


def load_pixium_frame_hdf5(path):
    """
    Load one 2D Pixium frame from .nxs/.hdf/.h5 using common I11 paths,
    then fallback to first 2D/3D dataset found.
    """
    known_paths = [
        "entry1/pixium_hdf/data",
        "/entry1/pixium_hdf/data",
        "entry/data/data",
        "/entry/data/data",
        "entry1/data/data",
        "/entry1/data/data",
        "entry1/Pixium10:detector/data",
        "/entry1/Pixium10:detector/data",
        "entry/Pixium10:detector/data",
        "/entry/Pixium10:detector/data",
        "entry1/detector/data",
        "/entry1/detector/data",
        "entry/detector/data",
        "/entry/detector/data",
    ]

    with h5py.File(path, "r") as f:
        for k in known_paths:
            if k in f:
                arr = np.array(f[k][()])
                if arr.ndim == 3:
                    return arr[0]
                if arr.ndim == 2:
                    return arr

        candidate = None

        def visit(name, obj):
            nonlocal candidate
            if isinstance(obj, h5py.Dataset):
                arr = np.array(obj[()])
                if arr.ndim == 2 and candidate is None:
                    candidate = arr
                    raise StopIteration
                if arr.ndim == 3 and candidate is None:
                    candidate = arr[0]

        try:
            f.visititems(visit)
        except StopIteration:
            pass

        if candidate is not None:
            return candidate

    raise RuntimeError(f"No suitable 2D Pixium dataset found in {path}")


def try_load_embedded_1d_hdf5(path):
    """
    Try to read embedded 1D arrays from HDF5/NeXus.
    Returns (two_theta, intensity) or raises ValueError if no usable pair exists.
    """
    with h5py.File(path, "r") as f:
        # Common guess paths from integrated data products
        guess_pairs = [
            ("entry1/integrated/two_theta", "entry1/integrated/intensity"),
            ("entry/integrated/two_theta", "entry/integrated/intensity"),
            ("entry1/results/two_theta", "entry1/results/intensity"),
            ("entry/results/two_theta", "entry/results/intensity"),
        ]
        for xk, yk in guess_pairs:
            if xk in f and yk in f:
                x = np.array(f[xk][()]).squeeze()
                y = np.array(f[yk][()]).squeeze()
                if x.ndim == 1 and y.ndim == 1 and len(x) == len(y) and len(x) > 10:
                    return x, y

    raise ValueError("No embedded 1D pattern found.")


def load_pattern_auto(path, two_theta_target=None, poni_path=None, mask_path=None, npt=3000):
    """
    Load pattern as 1D (2theta, intensity).
    Supported inputs: .xy/.xye/.dat/.txt/.nxs/.hdf/.h5
    """
    ext = os.path.splitext(path)[1].lower()

    if ext in (".xy", ".xye", ".dat", ".txt", ".csv"):
        data = robust_loadtxt(path, expected_cols=2)
        if data.ndim == 1:
            data = data.reshape(1, -1)
        if data.shape[1] < 2:
            raise ValueError(f"Text file {path} does not contain >=2 columns.")
        x = data[:, 0]
        y = data[:, 1]
        source_mode = "text-1d"

    elif ext in (".nxs", ".hdf", ".h5"):
        # 1) try direct embedded 1D first
        try:
            x, y = try_load_embedded_1d_hdf5(path)
            source_mode = "embedded-1d"
        except Exception:
            # 2) fallback to 2D integration with pyFAI
            resolved_poni = resolve_poni_path(poni_path)
            if resolved_poni is None:
                raise ValueError(
                    "FEP source is .nxs/.hdf but no embedded 1D pattern was found. "
                    "Provide a valid .poni path to integrate 2D -> 1D."
                )

            import pyFAI

            img = load_pixium_frame_hdf5(path).astype(np.float32, copy=False)
            ai = pyFAI.load(resolved_poni)

            mask = None
            if mask_path is not None and str(mask_path).strip() != "":
                if not os.path.exists(mask_path):
                    raise FileNotFoundError(f"Mask file not found: {mask_path}")
                mask = np.load(mask_path)

            npt_use = int(npt if npt is not None else 3000)
            res = ai.integrate1d(img, npt_use, mask=mask, unit="2th_deg")
            x = np.array(res[0]).astype(float)
            y = np.array(res[1]).astype(float)
            source_mode = "integrated-from-2d"
    else:
        raise ValueError(f"Unsupported FEP source extension: {ext}")

    # Ensure increasing x for interpolation
    order = np.argsort(x)
    x = x[order]
    y = y[order]

    if two_theta_target is not None:
        x_target = np.asarray(two_theta_target, dtype=float)
        y = np.interp(x_target, x, y)
        x = x_target

    return x, y, source_mode
```

## Gaussian Fit and FEP Removal 

```python
#FEP removal via Gaussian fitting 
from scipy.optimize import curve_fit

#Parameters
#this is incase the FEP peak is so intense that still causes issues even after automatoc subtraction - generally needed to get rid of one strong peak 
fep_peak_min = 5.6  # lower 2θ bound of FEP peak (adjust as needed)
fep_peak_max = 6.0  # upper 2θ bound of FEP peak (adjust as needed)



#1. Mask FEP region for fitting
window_mask = (two_theta_data >= fep_peak_min) & (two_theta_data <= fep_peak_max)
x_window = two_theta_data[window_mask]
y_window = intensity_data[window_mask]

#2. Define Gaussian function
def gauss(x, A, x0, sigma):
    return A * np.exp(-(x - x0)**2 / (2 * sigma**2))

#3. Initial guess for fitting parameters
A_guess = np.max(y_window)
x0_guess = x_window[np.argmax(y_window)]
sigma_guess = 0.05  # adjust if FEP peak is broader/narrower
p0 = [A_guess, x0_guess, sigma_guess]

#4. Fit Gaussian to FEP peak
try:
    popt, _ = curve_fit(gauss, x_window, y_window, p0=p0)
    print(f"Fitted FEP parameters: Amplitude={popt[0]:.2f}, Center={popt[1]:.4f}°, Sigma={popt[2]:.4f}°")
except RuntimeError:
    print("Gaussian fit failed; using raw FEP subtraction instead.")
    popt = None

#5. Construct fitted FEP over full 2θ range
if popt is not None:
    fep_fit_full = gauss(two_theta_data, *popt)
else:
    # fallback: use original intensity in window, zero elsewhere
    fep_fit_full = np.zeros_like(intensity_data)
    fep_fit_full[window_mask] = intensity_data[window_mask]

# 6. Subtract FEP from experimental data
data_minus_FEP = intensity_data - fep_fit_full
data_minus_FEP[data_minus_FEP < 0] = 0  # optional: clip negative intensities

# 7. Plot results
plt.figure(figsize=(10,5))
plt.plot(two_theta_data, intensity_data, label='Original data', alpha=0.5)
plt.plot(two_theta_data, fep_fit_full, label='Fitted FEP peak', lw=1)
plt.plot(two_theta_data, data_minus_FEP, label='Data - FEP', lw=1)
plt.legend()
plt.title("FEP removal via Gaussian fitting (weak peaks preserved)")
plt.xlabel("2θ (°)")
plt.ylabel("Intensity (a.u.)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

```

## Scaling and Single Graph Production 

```python
#detect main FEP peak region and safely compute local scaling

# 1. Use the Gaussian-fitted FEP from previous cell
FEP_for_scaling = fep_fit_full  # the fitted FEP

# Peak detection: fallback to simple argmax if scipy.find_peaks not available
if find_peaks is not None:
    threshold = np.max(FEP_for_scaling) * 0.5
    peaks, props = find_peaks(FEP_for_scaling, height=threshold)
    if len(peaks) == 0:
        peaks, props = find_peaks(FEP_for_scaling, height=np.max(FEP_for_scaling)*0.3)
else:
    peaks = np.array([np.argmax(FEP_for_scaling)])



# 2. Determine main peak index
if len(peaks) == 0:
    main_idx = int(np.argmax(FEP_for_scaling))
else:
    main_idx = peaks[np.argmax(FEP_for_scaling[peaks])]

peak_pos = two_theta_data[main_idx]
print(f"Detected main FEP peak at {peak_pos:.3f}° (index {main_idx})")



# 3. Define region width around main peak for scaling
region_width = 0.5  # degrees
mask_region = (two_theta_data >= (peak_pos - region_width)) & (two_theta_data <= (peak_pos + region_width))



# 4. Compute least-squares scale in the local region
num = np.sum(intensity_data[mask_region] * FEP_for_scaling[mask_region])
den = np.sum(FEP_for_scaling[mask_region] ** 2)
scale_local = (num / den) if den != 0 else 0.0
print(f"Local scale factor (region ±{region_width}°) = {scale_local:.6g}")



# 5. Subtract FEP only in the peak region and clip negatives
sub_local = intensity_data.copy()
sub_local[mask_region] -= FEP_for_scaling[mask_region] * scale_local
sub_local[sub_local < 0] = 0  # avoid negative intensities <<<--- this prevents the baseline from being negative 

# 6. Plot results
plt.figure(figsize=(10,6))
plt.plot(two_theta_data, intensity_data, label='Processed data', lw=1)
plt.plot(two_theta_data, FEP_for_scaling * scale_local, label=f'FEP scaled (s={scale_local:.4g})', lw=1)
plt.plot(two_theta_data, sub_local, label='Subtraction (clipped)', lw=1.2)
plt.xlim(peak_pos - 2.0, peak_pos + 2.0)
plt.legend()
plt.title("Local FEP subtraction (safe & clipped)")
plt.xlabel("2θ (°)")
plt.ylabel("Intensity (a.u.)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
#Apply global scale, subtract FEP, and safe baseline correction

# 1. manual scale is needed 
manual_scale = None   # <-- set a manual factor if needed, e.g., 0.95
scale_to_use = float(manual_scale) if manual_scale is not None else float(scale_local)
print(f"Using scale factor = {scale_to_use:.6g}")


# 2. Use Gaussian-fitted FEP instead of old shifted_FEP
scaled_fep_global = fep_fit_full * scale_to_use


# 3.  Subtract globally
subtracted_global = intensity_data - scaled_fep_global


# 4. baseline correction 
# Estimate baseline as the 5th percentile
baseline_shift = np.percentile(subtracted_global, 5)

# Shift pattern so baseline sits at ~0
subtracted_corrected = subtracted_global - baseline_shift

# Clip any negative values to 0 to avoid baseline going below zero
subtracted_corrected[subtracted_corrected < 0] = 0

print(f"Applied baseline shift = {baseline_shift:.4g} (clipped negatives to 0)")



# 5. Plot results 
plt.figure(figsize=(10,6))
plt.plot(two_theta_data, intensity_data, label='Processed data', alpha=0.7)
plt.plot(two_theta_data, scaled_fep_global, label='Scaled FEP (Gaussian fit)', alpha=0.7)
plt.plot(two_theta_data, subtracted_corrected, label='Final subtracted (baseline ~0)', lw=1.3, color='black')
plt.xlabel("2θ (°)")
plt.ylabel("Intensity (a.u.)")
plt.title("Final FEP subtraction (global application of local scale)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()



# 6. Save figure BEFORE show
out_png = os.path.join(processing_path, "FEP_subtracted_final.png")
plt.savefig(out_png, dpi=300)
plt.show()

print("Final: raw-sub min/max = {:.3g}/{:.3g}".format(subtracted_global.min(), subtracted_global.max()))
print("After baseline correction & clipping min/max = {:.3g}/{:.3g}".format(subtracted_corrected.min(), subtracted_corrected.max()))
print(f"Saved diagnostic figure to: {out_png}")

```
