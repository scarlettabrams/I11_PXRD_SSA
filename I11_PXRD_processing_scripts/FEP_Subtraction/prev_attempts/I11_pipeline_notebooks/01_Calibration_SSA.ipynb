```python
#Cell 1: imports and paths 
import os 
import json
import numpy as np 
import matplotlib.pyplot as plt 
import fabio 
import h5py 

from scipy.ndimage import median_filter 
from pyFAI.integrator.azimuthal import AzimuthalIntegrator 
```

```python
#Cell 2: loading the calibration files 
base_dir = r"E:/calib_test/" #load your directory - remember directory needs "/" not "\" and a "/" on the end to be read correctly 
calib_hdf = os.path.join(base_dir, "pixium_122543.hdf") #hdf = name of calibration file 

#after calibration, the poni and mask will be saved here 
out_dir = base_dir 
poni_out = os.path.join(out_dir, "i11_calib_test_CeO2.poni")
mask_out = os.path.join(out_dir, "hot_pixel_mask_test.npy")
```

```python
#Cell 3: HDF5 file Loader
def load_image_any(path):
    """Load a single 2D image froma file. Try Fabio first; if not, inspect HDF5."""
    path = os.path.abspath(path)

    #try fabio
    try:
        im = fabio.open(path)
        data = im.data.astype(np.float32)
        print(f"Loaded via fabio: {path} shape={data.shape}")
        return data 

    except Exception as e:
        print("fabio failed:", e, " - trying h5py fallback for HDF/nexus.")
    #try h5py: find a 2D dataset 

    with h5py.File(path, "r") as f:
        datasets = []
        def visitor(name, obj):
            try:
                if isinstance(obj, h5py.Dataset) and obj.ndim ==2:
                    datasets.append(name)
            except Exception:
                pass
        f.visititems(visitor)
        if not datasets:
            raise RuntimeError(f"no 2D dataset dound in {path} (checked with h5py).")
        #pick the first 2D dataset (usually correct for calibration images) 
        ds = datasets[0]
        data = f[ds][()].astype(np.float32)
        print(f"loaded via h5py dataset: {ds} shape={data.shape}")
        return data 
    
            
```

```python
#Cell 4: Show Calibration image 
img = load_image_any(calib_hdf)
plt.figure(figsize=(8,8))
plt.imshow(img, cmap="gray", vmax=np.percentile(img,99))
plt.title("Calibration image (preview)")
plt.colorbar()
plt.show()
```

```python
#Cell 6: load saved .poni and mask, then export paths for reuse
ai = None
mask = None

# Load .poni via AzimuthalIntegrator
if os.path.exists(poni_path):
    try:
        ai = AzimuthalIntegrator(poni_file=poni_path)
    except Exception:
        try:
            ai = AzimuthalIntegrator()
            ai.load(poni_path)
        except Exception as e:
            print("Failed to load .poni:", e)
else:
    print("No .poni found at:", poni_path)

# Load mask if present — search for .npy files in out_dir
mask_found_path = None
if os.path.exists(mask_out):
    # Try the expected path first
    try:
        mask = np.load(mask_out)
        mask_found_path = mask_out
        print(f"Loaded mask: {mask_out}, shape: {mask.shape}")
    except Exception as e:
        print(f"Failed to load mask at {mask_out}: {e}")
else:
    # Search for any .npy files in out_dir (pyfai may save with different name)
    try:
        npy_files = [f for f in os.listdir(out_dir) if f.endswith('.npy')]
        if npy_files:
            mask_candidate = os.path.join(out_dir, npy_files[0])
            mask = np.load(mask_candidate)
            mask_found_path = mask_candidate
            print(f"Loaded mask: {mask_candidate}, shape: {mask.shape}")
        else:
            print("No .npy mask file found in:", out_dir)
    except Exception as e:
        print(f"Error searching for mask: {e}")

# Summarize calibration
if ai is not None:
    print(f"✓ Successfully loaded .poni: {poni_path}")
    try:
        print(f"   Detector distance (m): {ai.dist:.4f}")
        print(f"   Beam center (px): poni1={ai.poni1:.2f}, poni2={ai.poni2:.2f}")
    except Exception:
        pass
else:
    print("✗ AzimuthalIntegrator not loaded — check .poni path.")

if mask is None:
    print("⚠ No mask loaded (optional).")

# Export paths for other notebooks
config_path = os.path.join(out_dir, "calibration_paths.json")
try:
    mask_path_to_save = mask_found_path if mask_found_path else mask_out
    with open(config_path, "w") as f:
        json.dump({
            "poni_path": os.path.abspath(poni_path),
            "mask_path": os.path.abspath(mask_path_to_save)
        }, f, indent=2)
    print(f"✓ Saved calibration paths to: {config_path}")
except Exception as e:
    print("Failed to write calibration_paths.json:", e)
```

```python
#Cell 7: Verify calibration is complete and correct
print("\n=== CALIBRATION VERIFICATION ===\n")

# Check that both ai and mask loaded
if ai is None:
    print("ERROR: AzimuthalIntegrator not loaded. Check .poni file exists and is valid.")
else:
    print("✓ AzimuthalIntegrator loaded successfully")
    try:
        print(f"  Detector distance: {ai.dist:.4f} m")
        print(f"  Beam center (poni1, poni2): ({ai.poni1:.2f}, {ai.poni2:.2f}) px")
        print(f"  Wavelength: {ai.wavelength*1e10:.4f} Å")
    except Exception as e:
        print(f"  (Could not extract all parameters: {e})")

if mask is None:
    print("⚠ No mask loaded (optional).")
else:
    print(f"✓ Mask loaded: shape {mask.shape}, {mask.sum()} masked pixels")

# Test integration on calibration image
print("\nRunning test 1D integration on calibration image...")
try:
    npt = 1500
    res = ai.integrate1d(img.astype(np.float32), npt, mask=mask, unit="2th_deg")
    
    # Extract radial and intensity (handle both tuple and object returns)
    try:
        radial = res.radial
        intensity = res.intensity
    except AttributeError:
        radial = res[0]
        intensity = res[1]
    
    print(f"✓ Integration successful: {len(radial)} points")
    
    # Plot to visually confirm calibration
    plt.figure(figsize=(10, 5))
    plt.plot(radial, intensity, 'b-', linewidth=1)
    plt.xlabel("2θ (deg)", fontsize=12)
    plt.ylabel("Intensity", fontsize=12)
    plt.title("Test 1D Integration - Calibration Image", fontsize=12)
    plt.grid(alpha=0.3)
    
    # Save plot as PNG
    png_path = os.path.join(out_dir, "calibration_integration_test.png")
    plt.savefig(png_path, dpi=150, bbox_inches='tight')
    print(f"✓ Saved plot to: {png_path}")
    plt.show()
    
    # Save radial and intensity as .xy file
    xy_path = os.path.join(out_dir, "calibration_integration_test.xy")
    np.savetxt(xy_path, np.column_stack([radial, intensity]), 
               header="2theta(deg) Intensity", delimiter="\t", fmt="%.6f")
    print(f"✓ Saved integration data to: {xy_path}")
    
    print("✓ Calibration appears correct — rings/peaks should be visible above.")
    
except Exception as e:
    print(f"✗ Integration failed: {e}")
    print("  This may indicate a problem with the .poni file or mask.")

print("\n=== END VERIFICATION ===\n")
```
