# Baseline correction 

### using https://pybaselines.readthedocs.io/en/latest/ mor() baseline correction

```python

intensity_values = oneD_integrated_plot[1]
two_theta = oneD_integrated_plot[0]
x = two_theta
y = intensity_values

baseline_fitter = Baseline(x_data=x)
```

```python
len(oneD_integrated_plot)
```

```python
#scarlett
#Determine Half window value :)
# typically PXRD Data uses the range of 5-15 
# You want the baseline to follow the general background trend without cutting into any true diffraction peaks.
plt.figure(figsize=(10,6))
for hw in [3, 7, 10, 15]:
    baseline = baseline_fitter.mor(y, half_window=hw)[0]
    plt.plot(x, baseline, label=f'half_window={hw}')
plt.plot(x, y, 'k--', alpha=0.3, label='data')
plt.legend()
plt.xlabel('2θ (°)')
plt.ylabel('Intensity')
plt.title('Effect of half_window on baseline shape')
plt.show()
```

```python
#Scarlett 
#automated estimate of suitable half window value 
# Calculate step size in 2θ
step_size = np.mean(np.diff(x))  # degrees per data point

# Estimate how many data points correspond to ~0.3–0.5° of 2θ background curvature
# (adjust this range if your background varies faster or slower)
background_scale_deg = 0.4  # typical slow-varying background over ~0.4°
half_window_estimate = int(background_scale_deg / step_size / 2)

# Ensure it's an odd, reasonable value (avoid too small or too large)
half_window = max(3, min(half_window_estimate, 25))

print(f"Estimated half_window = {half_window} (based on step size {step_size:.4f}°)")
```

```python
# baseline correction, change half_window valuue to best fit dataset

half_window = 6

plt.figure()
plt.plot(x, y, label='data')
plt.plot(x, baseline_fitter.mor(y, half_window=half_window)[0], label=f'half_window={half_window}')
plt.legend()
```

```python
# baseline correction
corrected_data = y - baseline_fitter.mor(y, half_window=half_window)[0]
```

```python
# plotting of baseline corrected pattern against baseline correction parameters

twoD_pattern_name_baseline_corr = data_file_no +"_baseline_corr_half_window_" + str(half_window)

baseline_corr_name = data_file_no + "half_window_" + str(half_window) + "_baseline"

plt.figure()
plt.plot(x, y, label='data')
plt.plot(x, baseline_fitter.mor(y, half_window=half_window)[0], label=f'half_window={half_window}')
plt.plot(x, corrected_data, label='corrected_baseline')
plt.title(baseline_corr_name, fontsize=15)
plt.xlabel('$2\\theta$ ($^{o}$)')
plt.ylabel('Intensity')
plt.xlim(4, 20)
plt.legend()
plt.savefig(processing_path + twoD_pattern_name_baseline_corr)
```

```python
# final pattern

plt.figure()
plt.plot(x, corrected_data, label= 'final_pattern')
plt.title(twoD_pattern_name_baseline_corr, fontsize=15)
plt.xlabel('$2\\theta$ ($^{o}$)')
plt.ylabel('Intensity')
plt.xlim(4, 30)
#plt.ylim(-10, 2000)
plt.savefig(processing_path + twoD_pattern_name_baseline_corr + "_final_pattern")
```

```python
# exporting final pattern as .xy data

np.savetxt(processing_path+twoD_pattern_name_baseline_corr+ ".xy", np.c_[x, corrected_data])
```

```python
# batch baseline correction for all diffraction frames in data_file_no_list
half_window_batch = half_window if 'half_window' in globals() else 6
file_numbers = data_file_no_list if 'data_file_no_list' in globals() else [int(data_file_no)]

processed = []
failed = []

for file_no in file_numbers:
    file_no = str(file_no)
    data_file_i = f"i11-1-{file_no}.nxs"
    candidate_paths_i = [os.path.join(data_dir, data_file_i)]
    candidate_paths_i += [
        os.path.join(data_dir, d, data_file_i)
        for d in os.listdir(data_dir)
        if os.path.isdir(os.path.join(data_dir, d))
    ]
    file_path_i = next((p for p in candidate_paths_i if os.path.exists(p)), None)

    if file_path_i is None:
        failed.append((file_no, 'file not found'))
        continue

    try:
        with h5py.File(file_path_i, 'r') as dat:
            myData_i = np.array(dat['/entry1/pixium_hdf/data'][()][:])
            diff_frame_i = myData_i.reshape(myData_i.shape[1:])

        oneD_integrated_plot_i = calibration.integrate1d(
            diff_frame_i,
            1000,
            unit=pyFAI.units.TTH_DEG,
            radial_range=[0, 30],
            mask=oneD_mask
)

        x_i = oneD_integrated_plot_i[0]
        y_i = oneD_integrated_plot_i[1]
        baseline_i = Baseline(x_data=x_i).mor(y_i, half_window=half_window_batch)[0]
        corrected_data_i = y_i - baseline_i

        twoD_pattern_name_baseline_corr_i = f"{file_no}_baseline_corr_half_window_{half_window_batch}"
        baseline_corr_name_i = f"{file_no}half_window_{half_window_batch}_baseline"

        plt.figure()
        plt.plot(x_i, y_i, label='data')
        plt.plot(x_i, baseline_i, label=f'half_window={half_window_batch}')
        plt.plot(x_i, corrected_data_i, label='corrected_baseline')
        plt.title(baseline_corr_name_i, fontsize=15)
        plt.xlabel('$2\\theta$ ($^{o}$)')
        plt.ylabel('Intensity')
        plt.xlim(4, 20)
        plt.legend()
        plt.savefig(os.path.join(processing_path, twoD_pattern_name_baseline_corr_i))
        plt.close()

        plt.figure()
        plt.plot(x_i, corrected_data_i, label='final_pattern')
        plt.title(twoD_pattern_name_baseline_corr_i, fontsize=15)
        plt.xlabel('$2\\theta$ ($^{o}$)')
        plt.ylabel('Intensity')
        plt.xlim(4, 30)
        plt.savefig(os.path.join(processing_path, twoD_pattern_name_baseline_corr_i + '_final_pattern'))
        plt.close()

        np.savetxt(
            os.path.join(processing_path, twoD_pattern_name_baseline_corr_i + '.xy'),
            np.c_[x_i, corrected_data_i]
        )

        processed.append(file_no)
    except Exception as e:
        failed.append((file_no, str(e)))

print(f'Processed {len(processed)} files; failed {len(failed)} files.')
if failed:
    print('First 10 failures:')
    for item in failed[:10]:
        print(item)
```
