##  Read in Background Files - do once

```python
#scarlett adaptations 
# read in background frames and calculate total counts - be patient - can take a while with large background sets


bg_file_nxs = [f for f in os.listdir(background_dir) if f.endswith(".nxs")]
count = len(bg_file_nxs)

# hard-coded detector size
rows, cols = 2881, 2880

bg_frame_sized = np.zeros((count, rows, cols))
basic_bg_sub   = np.zeros((count, rows, cols))
total_intensity = np.zeros(count)

for a in range(count):
    with h5py.File(os.path.join(background_dir, bg_file_nxs[a]), 'r') as dat:
        background_frame = np.array(dat["/entry1/pixium_hdf/data"][()])
        bg_frame_sized[a] = background_frame.reshape(rows, cols)
        basic_bg_sub[a] = np.subtract(diff_frame, bg_frame_sized[a])
        basic_bg_sub[a][basic_bg_sub[a] < 0] = 0
        total_intensity[a] = np.sum(basic_bg_sub[a])
```

## Few Backgorund Data sets - Under 40

```python
# 40 or less backgorund frames recommended for visualisation
# LARGE MEMORY USAGE 
# visualisation of background best fit checks. 

import math

cols = 3
rows = math.ceil(count / cols)

fig, axs = plt.subplots(rows, cols, figsize=(15, rows*5), facecolor='w', edgecolor='k')
fig.subplots_adjust(hspace=.5, wspace=.001)
axs = axs.ravel()

for i in range(count):
    fig.suptitle('Basic background subtraction best fit check', fontsize=24)
    axs[i].imshow(basic_bg_sub[i], cmap='gray', vmin=0, vmax=100)
    axs[i].set_title(f"{bg_file_nxs[i]}\nTotal intensity: {round(total_intensity[i])}",
                     fontweight='bold')

# hide any unused subplots if rows*cols > count
for j in range(count, len(axs)):
    fig.delaxes(axs[j])
```

## Large Background Data sets - Over 40 

```python
#  over 40 background frames!!!! 
# ONLY RUN IF PREVIOUS CELL RAISED MEMORYERROR
# # LOW-MEMORY FALLBACK (run this instead if MemoryError occurs above)
# Use float math to avoid unsigned-integer underflow during subtraction.

import math

bg_file_nxs = [f for f in os.listdir(background_dir) if f.endswith('.nxs')]
count = len(bg_file_nxs)
rows, cols = diff_frame.shape

# force float subtraction (critical for correct background scoring)
diff_frame_f = diff_frame.astype(np.float32, copy=False)
total_intensity = np.zeros(count, dtype=np.float64)

for a, fname in enumerate(bg_file_nxs):
    with h5py.File(os.path.join(background_dir, fname), 'r') as dat:
        background_frame = np.array(dat['/entry1/pixium_hdf/data'][()]).reshape(rows, cols).astype(np.float32, copy=False)

    bg_sub = diff_frame_f - background_frame
    bg_sub[bg_sub < 0] = 0
    total_intensity[a] = np.sum(bg_sub, dtype=np.float64)

# Plot only best candidates to keep plotting manageable
max_plots = min(12, count)
best_idx = np.argsort(total_intensity)[:max_plots]

plot_cols = 3
plot_rows = math.ceil(max_plots / plot_cols)
fig, axs = plt.subplots(plot_rows, plot_cols, figsize=(15, plot_rows * 5), facecolor='w', edgecolor='k')
fig.subplots_adjust(hspace=.5, wspace=.05)
axs = np.atleast_1d(axs).ravel()
fig.suptitle('Low-memory background best-fit check', fontsize=24)

for i, bg_idx in enumerate(best_idx):
    with h5py.File(os.path.join(background_dir, bg_file_nxs[bg_idx]), 'r') as dat:
        background_frame = np.array(dat['/entry1/pixium_hdf/data'][()]).reshape(rows, cols).astype(np.float32, copy=False)
    bg_sub = diff_frame_f - background_frame
    bg_sub[bg_sub < 0] = 0

    axs[i].imshow(bg_sub, cmap='gray', vmin=0, vmax=100)
    axs[i].set_title(
        f"{bg_file_nxs[bg_idx]}\nTotal intensity: {round(total_intensity[bg_idx])}",
        fontweight='bold'
    )

for j in range(max_plots, len(axs)):
    fig.delaxes(axs[j])

# Auto-select best background and load it for downstream cells
idx = int(np.argmin(total_intensity))
chosen_bg_name = bg_file_nxs[idx]
with h5py.File(os.path.join(background_dir, chosen_bg_name), 'r') as dat:
    chosen_bg = np.array(dat['/entry1/pixium_hdf/data'][()]).reshape(rows, cols).astype(np.float32, copy=False)

print(f"Low-memory mode: scored {count} files")
print(f"Chosen background: {chosen_bg_name}")
print(f"Total intensity: {total_intensity[idx]}")
print("Run the next subtraction cell as normal.")
```

## Background File Selection

```python
# Scarlett adaptation for manual overwrite of background selection (auto by default)
auto_idx = int(np.argmin(total_intensity))

# --- manual overwrite by collection number in filename ---
# Example: manual_collection_no = "122815"
manual_collection_no = None # keep None for automatic selection

if manual_collection_no is None:
    idx = auto_idx
else:
    manual_collection_no = str(manual_collection_no).strip()
    if not manual_collection_no:
        raise ValueError("manual_collection_no cannot be empty. Use None for automatic selection.")

    # Match background filenames containing the given collection number
    matching_idx = [i for i, fname in enumerate(bg_file_nxs) if manual_collection_no in str(fname)]
    if len(matching_idx) == 0:
        raise ValueError(
            f"No background file contains collection number '{manual_collection_no}'. "
            f"Available files: {bg_file_nxs}"
        )
    if len(matching_idx) > 1:
        matches = [bg_file_nxs[i] for i in matching_idx]
        raise ValueError(
            f"Multiple background files match collection number '{manual_collection_no}': {matches}. "
            "Please provide a more specific collection number."
)

    idx = int(matching_idx[0])

chosen_bg_name = bg_file_nxs[idx]

# Works in both modes:
# - full-memory workflow: reuse preloaded bg_frame_sized
# - low-memory workflow: load only the chosen frame from disk
if 'bg_frame_sized' in globals() and isinstance(bg_frame_sized, np.ndarray) and bg_frame_sized.shape[0] > idx:
    chosen_bg = bg_frame_sized[idx].astype(np.float32, copy=False)
else:
    with h5py.File(os.path.join(background_dir, chosen_bg_name), 'r') as dat:
        rows, cols = diff_frame.shape
        chosen_bg = np.array(dat['/entry1/pixium_hdf/data'][()]).reshape(rows, cols).astype(np.float32, copy=False)

print('Chosen background: ' + str(chosen_bg_name) +
      '\nTotal intensity: ' + str(total_intensity[idx]))
```

# Thresholding 

now we have a better 2D pattern, try seperate a bit of the signal from noise in the diffraction pattern 

```python
#beth 
# thresholding to remove remaining noise - change threshold_value

pattern_processing = diff_frame - chosen_bg

threshold_value = 70 #120 was the value when Beth handed code over just fyi :) - higher the value - more 'masking'out of noise (been finding 70 has been working generaly) 

pattern_processing[pattern_processing < threshold_value] = 0

thresholded_frame = pattern_processing
```

# Integration using pyFAI module 

### Fast Azimuthal Integration using Python - pyFAI: https://pyfai.readthedocs.io/en/v2023.1/

core idea is to redistribute the signal acquired with the experimental geometry into a geometry suitable for further analysis - e.g. Reitveld refinement 

```python
# integrate thresholded frame using calibration + oneD_mask already loaded above

# this produces 1D array data Vs 2theta ready for plotting 
oneD_integrated_plot_thresholded = calibration.integrate1d(
    thresholded_frame,
    1000,
    unit=pyFAI.units.TTH_DEG,
    radial_range=[0,30],
    mask=oneD_mask
)
```

```python
# Re-threshold if noise remains
# plot 
jupyter.plot1d(oneD_integrated_plot_thresholded)
plt.title(threshold_pattern_name, fontsize=15)
plt.xlabel('$2\\theta$ ($^{o}$)')
plt.xlim(4, 30)
#plt.ylim(-0.1, 40)
plt.ylabel('Intensity')
#plt.savefig(processing_path + threshold_pattern_name +"_1D_plot_zoom")
```

```python
# Scarlett - Save 1D thresholded plot, XY, and XYE data together

# Uses threshold_save_dir defined in calibration setup cell above
os.makedirs(threshold_save_dir, exist_ok=True)

#1D plot save as png 
plt.figure(figsize=(10,6))
plt.plot(oneD_integrated_plot_thresholded.radial, oneD_integrated_plot_thresholded.intensity, color='black')
plt.title(threshold_pattern_name, fontsize=15)
plt.xlabel('$2\\theta$ ($^{o}$)')
plt.ylabel('Intensity')
plt.xlim(4, 30)

png_path = os.path.join(threshold_save_dir, threshold_pattern_name + "_1D_plot.png")
plt.savefig(png_path, dpi=300, bbox_inches="tight")
plt.close()

# Extract arrays
x_vals = oneD_integrated_plot_thresholded.radial   # 2θ
y_vals = oneD_integrated_plot_thresholded.intensity  # Intensity
errors = np.sqrt(np.maximum(y_vals, 1))  # simple Poisson estimate

#1D plt save as .xy
xy_path = os.path.join(threshold_save_dir, threshold_pattern_name + ".xy")
np.savetxt(xy_path, np.c_[x_vals, y_vals],
           fmt="%.6f", header="2theta(deg) Intensity", comments='')

#1D plot save as .xye
xye_path = os.path.join(threshold_save_dir, threshold_pattern_name + ".xye")
np.savetxt(xye_path, np.c_[x_vals, y_vals, errors],
           fmt="%.6f", header="2theta(deg) Intensity Error", comments='')

print(f"Saved plot to: {png_path}")
print(f"Saved XY data to: {xy_path}")
print(f"Saved XYE data to: {xye_path}")
```
