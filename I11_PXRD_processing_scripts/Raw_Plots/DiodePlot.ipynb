# Diode `.dat` Plot Export

Load a diode experiment `.dat` file, plot it, and save the figure as a `.png` in a directory you choose.

```python
import os
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
```

## Configure paths

Set the exact `.dat` file path and the directory where the PNG should be written.

```python
# Set these two paths before running the notebook.
input_dat_path = Path(r"E:/I11 Beamtime July 25 - LHIST/RAW_2D/Diode plots/107798_107801_diode.dat/")
output_dir = Path(r"C:/Users/ezxsa27/OneDrive - The University of Nottingham/Nottingham/Year 2 (Jul 25 -)/Yr2 Progression Review/")

output_dir.mkdir(parents=True, exist_ok=True)
output_png_path = output_dir / f"{input_dat_path.stem}.png"

if not input_dat_path.exists():
    raise FileNotFoundError(f"Input file not found: {input_dat_path}")
```

```python
def load_two_column_dat(filepath):
    values = []
    with open(filepath, "r", encoding="utf-8", errors="ignore") as handle:
        for line in handle:
            stripped = line.strip()
            if not stripped or stripped.startswith("#") or stripped.startswith(";"):
                continue
            parts = stripped.replace(",", " ").split()
            if len(parts) < 2:
                continue
            try:
                values.append((float(parts[0]), float(parts[1])))
            except ValueError:
                continue
    data = np.asarray(values, dtype=float)
    if data.size == 0:
        raise ValueError(f"No numeric 2-column data found in {filepath}")
    return data[:, 0], data[:, 1]
```

```python
x, y = load_two_column_dat(input_dat_path)

plt.figure(figsize=(8, 4.5))
plt.plot(x, y, linewidth=1.2, color="black")
#plt.xlabel("X")
plt.xlim(np.min(30), np.max(35))
plt.xticks([])
plt.ylabel("Intensity")
plt.title("Diode Plot") #(input_dat_path.stem)
plt.tight_layout()
plt.savefig(output_png_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"Loaded {len(x)} points from {input_dat_path}")
print(f"Saved PNG to {output_png_path}")
```
